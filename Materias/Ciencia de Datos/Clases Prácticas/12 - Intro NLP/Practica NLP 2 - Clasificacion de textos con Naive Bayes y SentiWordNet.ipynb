## Ley de Zipf

Antes de extraer features conviene mirar cómo se distribuyen las palabras del corpus, porque esa distribución justifica varias de las decisiones que vamos a tomar a continuación.

La **ley de Zipf** dice que la frecuencia de una palabra es **inversamente proporcional a su rango** en la tabla de frecuencias: la palabra más frecuente (rango 1) aparece aproximadamente el doble de veces que la segunda (rango 2), el triple que la tercera (rango 3), y así sucesivamente. Formalmente:

$$\text{freq}(w) \;\propto\; \frac{1}{\text{rango}(w)}$$

La consecuencia práctica es que el vocabulario de cualquier corpus tiene **un puñado de palabras hiperfrecuentes que dominan el texto y una cola larguísima de palabras rarísimas**. Veámoslo en nuestros datos.

```python
# Series con todas las palabras del corpus
words_series = pd.Series([word for rev in df.review for word in word_tokenize(rev) if word.isalpha()])
```

```python
# Frecuencia de cada palabra en el corpus
word_freq = words_series.value_counts()
word_freq
```

```python
# plot de frecuencia vs ranking (la palabra mas frecuente tiene rank=1, la segunda tiene rank=2, etc..)
plt.plot(range(len(word_freq)),word_freq)
plt.xlabel('rank');plt.ylabel('freq');
```

En escala **log-log**, una distribución que cumple Zipf se ve aproximadamente como una **recta de pendiente ≈ −1**. Por eso volvemos a graficar lo mismo pero con ambos ejes en escala logarítmica: si la nube de puntos se alinea sobre una recta descendente, estamos viendo la ley de Zipf en acción.


Los dos números de arriba no son una curiosidad: explican directamente cómo vamos a construir las features.

- **El umbral de frecuencia mínima (`min_df`).** Ese ~35% de palabras que aparecen una sola vez (*hapax legomena*) es la cola de Zipf. No aportan señal estadística —no se puede estimar $P(w \mid c)$ de forma confiable con una única observación— y solo agregan ruido y dimensionalidad. Cuando más abajo usemos `min_df=5` en el `CountVectorizer`, estaremos literalmente **podando esta cola**.

- **Las stopwords.** Las palabras del extremo opuesto —ese puñado que concentra una porción enorme del corpus— son justamente las que casi no discriminan entre clases ('the', 'a', 'of'...). Filtrarlas con un `stoplist` es **recortar la cabeza de la curva de Zipf**.

- **La esparsidad (*sparsity*) de la matriz de features.** Como casi todas las palabras del vocabulario son raras, cualquier documento individual contiene solo una fracción minúscula del vocabulario total. Por eso la matriz documento-término que generaremos será **altamente dispersa** (mayoría de ceros), algo que mediremos explícitamente unas celdas más adelante.

En resumen: Zipf no es un apéndice del análisis de texto, es la razón de fondo por la que el preprocesamiento de NLP se ve como se ve.

# Naive Bayes

**Naive Bayes** es un clasificador probabilístico clásico para texto (por ejemplo, clasificar un documento como *positivo*/*negativo*, o por tema). Se apoya en el **teorema de Bayes** y opera muy bien sobre representaciones de bolsa de palabras.

La idea es: dado un documento $d$ (una bolsa de palabras), ¿cuál es la clase $c$ más probable? Buscamos la clase que **maximiza** la probabilidad a posteriori:

$$\hat{c} = \arg\max_{c \in C} \; P(c \mid d)$$

Por el **teorema de Bayes**, esa probabilidad se reescribe como:

$$P(c \mid d) = \frac{P(d \mid c)\, P(c)}{P(d)}$$

Como $P(d)$ es igual para todas las clases (no afecta cuál es el máximo), podemos ignorarlo:

$$\hat{c} = \arg\max_{c \in C} \; \underbrace{P(d \mid c)}_{\text{verosimilitud}} \; \underbrace{P(c)}_{\text{prior}}$$

Acá entra el supuesto **"naive" (ingenuo)**: asumimos que las palabras del documento son **condicionalmente independientes** dada la clase. Es decir, la presencia de una palabra no depende de las demás. Es falso en el lenguaje real, pero simplifica enormemente el cálculo y, en la práctica, funciona *bastante* bien. Esto nos deja multiplicar la probabilidad de cada palabra $w_i$:

$$\hat{c} = \arg\max_{c \in C} \; P(c) \prod_{i=1}^{n} P(w_i \mid c)$$

donde $P(w_i \mid c)$ es la probabilidad de que aparezca la palabra $w_i$ en documentos de la clase $c$, estimada simplemente **contando** sobre el corpus de entrenamiento.

### En la práctica: trabajamos con logaritmos

Multiplicar muchas probabilidades chicas lleva a *underflow* (números tan pequeños que la computadora los redondea a cero). Por eso se aplica $\log$, que convierte el producto en suma y es **monótono** (no cambia cuál es el máximo):

$$\hat{c} = \arg\max_{c \in C} \; \log P(c) + \sum_{i=1}^{n} \log P(w_i \mid c)$$

### Suavizado de Laplace

Si una palabra nunca apareció en una clase durante el entrenamiento, $P(w_i \mid c) = 0$ y **anula todo el producto** (un solo cero arruina la predicción). Para evitarlo se suma una constante (típicamente 1) a cada conteo, lo que se conoce como **suavizado de Laplace** (*add-one smoothing*):

$$P(w_i \mid c) = \frac{\text{count}(w_i, c) + 1}{\sum_{w \in V}\big(\text{count}(w, c) + 1\big)} = \frac{\text{count}(w_i, c) + 1}{\Big(\sum_{w \in V}\text{count}(w, c)\Big) + |V|}$$

donde $|V|$ es el tamaño del vocabulario.

```python
clf = MultinomialNB(alpha=1) # alpha es el smoothing parameter
scores_cv_nb = cross_val_score(clf,
                               X_train, y_train,
                               cv = 10, scoring='f1_macro')
scores_cv_nb
```

### Dummy

El `DummyClassifier` **ignora deliberadamente las features**: no mira el contenido de los textos, predice usando una regla trivial. No sirve para clasificar en serio, sino para darnos un **piso** contra el cual comparar. Si nuestro modelo no le gana al dummy, entonces no aprendió nada útil.

Posee distintas estrategias principales (`strategy=`), entre ellas
`stratified` (predice al azar respetando las proporciones de clase del entrenamiento), `most_frequent` (predice siempre la clase mayoritaria), `uniform` (predice al azar de forma uniforme entre todas las clases, entre otras.

**¿Por qué importa tanto? Porque un número de performance no significa nada en abstracto.**

Nuestro corpus está balanceado (50% pos / 50% neg), así que el dummy `stratified` saca un f-score cercano a 0.50. Ahora imaginen un problema donde el 95% de los casos son de una sola clase. ¿Qué accuracy obtendría un dummy `most_frequent` *sin aprender nada*? ¿Sigue siendo la accuracy una buena métrica en ese escenario? ¿Por qué un mismo número (digamos, 0.80) puede ser excelente en un problema y mediocre en otro?

# Lexicon (SentiWordNet)

El enfoque léxico es **no supervisado** (en el sentido de que no entrena con etiquetas). Usa un diccionario de sentimiento ya construido por humanos/expertos —el lexicon— donde cada palabra ya viene con un puntaje de polaridad. No mira las etiquetas del corpus para nada; podría clasificar reviews sin haber visto jamás una sola review etiquetada.

WordNet es una base de datos léxica del inglés. Agrupa palabras en *synsets* (conjuntos de sinónimos), donde cada synset representa un concepto/sentido. Una palabra polisémica pertenece a varios synsets. Por ejemplo "bank" tiene un synset para la orilla del río y otro para la institución financiera.

SentiWordNet le pega encima a cada synset de WordNet tres puntajes: positivo, negativo y objetivo (neutral), que suman 1. Así, el synset financiero de "bank" puede ser casi 100% objetivo, mientras que el synset de "awful" tendrá un negativo alto.

SentiWordNet complementa a WordNet al asignar puntajes numéricos que indican la polaridad de sentimiento de cada synset. Estos puntajes son:

1.   Puntaje Positivo: Indica cuán positivo es el sentimiento del synset (entre 0 y 1).
2.  Puntaje Negativo: Indica cuán negativo es el sentimiento del synset (entre 0 y 1).
3. Puntaje Objetivo: Indica cuán objetivo u neutral es el sentimiento del synset (entre 0 y 1).

La suma de los tres puntajes para un synset siempre es 1.



```python
# Ejemplo: obtener puntajes de sentimiento para la palabra 'awful'
word = wn.synsets('awful')[0] # Obtenemos el primer synset de 'awful'
sentiment_scores = swn.senti_synset(word.name()) # Obtenemos los puntajes de sentimiento

print(f"Palabra: {word.name()}")
print(f"Scores: {sentiment_scores}")
print(f"Puntaje Positivo: {sentiment_scores.pos_score()}")
print(f"Puntaje Negativo: {sentiment_scores.neg_score()}")
print(f"Puntaje Objetivo: {sentiment_scores.obj_score()}")
```

A continuación, armaremos un clasificador que hace lo siguiente, palabra por palabra de cada review (`get_score`):

* Tokeniza la oración/review.
* Para cada palabra, busca su synset en SentiWordNet y extrae `pos_score` y `neg_score`.
* Acumula esos puntajes (sumándolos o promediándolos según method).
* Calcula un score final = `pos_total − neg_total`.
* Si ese score supera un umbral (threshold), clasifica como "pos"; si no, "neg".


```python
def penn_to_wn(tag):
    """
    Convert between the PennTreebank tags to simple Wordnet tags
    """
    if tag.startswith('J'):
        return wn.ADJ
    elif tag.startswith('N'):
        return wn.NOUN
    elif tag.startswith('R'):
        return wn.ADV
    elif tag.startswith('V'):
        return wn.VERB
    return None


def get_word_sentiment(word , wn_tag_bool=False):
    """
    Función para obtener el puntaje de sentimiento de una palabra
    """
    if wn_tag_bool:
      wn_tag = penn_to_wn(word)
      if wn_tag not in (wn.NOUN, wn.ADJ, wn.ADV):
          return []

    sentiment_scores = list(swn.senti_synsets(word, pos=wn_tag))  if wn_tag_bool else list(swn.senti_synsets(word))
    if sentiment_scores:
        return sentiment_scores[0]
    else:
        return None

def get_score(sentence, method='avg'):
    """
    Función para clasificar una oración como positiva o negativa
    """
    words = nltk.word_tokenize(sentence)
    pos_scores = []
    neg_scores = []
    for word in words:
        sentiment = get_word_sentiment(word)
        # print(f'word: {word} \t sentiment: {sentiment}')
        if sentiment:
            pos_scores.append(sentiment.pos_score())
            neg_scores.append(sentiment.neg_score())

    if method == 'avg':
      pos_score = sum(pos_scores) / len(pos_scores) if pos_scores else 0
      neg_score = sum(neg_scores) / len(neg_scores) if neg_scores else 0
    elif method == 'sum':
      pos_score = sum(pos_scores)  if pos_scores else 0
      neg_score = sum(neg_scores)  if neg_scores else 0

    return pos_score - neg_score

def classify_sentence(score, threshold=0.5):
  if score >= threshold:
        return 'pos'
  else:
        return 'neg'
```

```python
def make_prediction(X_to_predict, method='avg', threshold=0.5 ):
  """
  Clasifica los datos
  """
  y_scores = []
  y_predict = []
  for sentence in X_to_predict:
      score = get_score(sentence, method)
      label = classify_sentence(score, threshold )
      y_scores.append(score)
      y_predict.append(label)

  return y_predict, y_scores
```

```python
def get_metrics(y_predict, y_ground_truth):
    """
    Evaluar en los datos de prueba
    """
    correct = 0
    total = 0
    true_positive = 0
    false_positive = 0
    false_negative = 0

    for predicted_label, true_label in zip(y_predict, y_ground_truth):
        if predicted_label == true_label:
            correct += 1
            if predicted_label == 'pos':  # Assuming binary classification with labels 0 and 1
                true_positive += 1
        else:
            if predicted_label == 'pos':
                false_positive += 1
            if true_label == 'pos':
                false_negative += 1
        total += 1

    accuracy = correct / total

    if true_positive + false_positive > 0:
        precision = true_positive / (true_positive + false_positive)
    else:
        precision = 0

    if true_positive + false_negative > 0:
        recall = true_positive / (true_positive + false_negative)
    else:
        recall = 0

    if precision + recall > 0:
        f1_score = 2 * (precision * recall) / (precision + recall)
    else:
        f1_score = 0

    print(f"Accuracy: {accuracy * 100:.2f}%")
    print(f"Precision: {precision*100:.2f}%")
    print(f"Recall: {recall*100:.2f}%")
    print(f"F1 Score: {f1_score*100:.2f}%")
```

```python
y_train_predict, y_train_scores = make_prediction(X_train_text, method='sum', threshold=10)
y_test_predict, y_test_scores = make_prediction(X_test_text, method='sum', threshold=10)
```
